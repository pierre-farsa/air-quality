# Filtering weather dataset

```python
df_weather = pd.read_csv(Path("../data/raw/Q_75_previous-1950-2024_RR-T-Vent.csv"), sep=";")
df_weather_ap = pd.read_csv(Path("../data/raw/Q_75_previous-1950-2024_autres-parametres.csv"), sep=";")
```

```python
# Filtering on column AAAAMMJJ for years 2021 to 2023
df_weather = df_weather[df_weather["AAAAMMJJ"].between(20210101, 20231231)]
df_weather_ap = df_weather_ap[df_weather_ap["AAAAMMJJ"].between(20210101, 20231231)]
```

```python
# Writing filtered weather data to new CSV files
# df_weather.to_csv(Path("../data/processed/data_weather.csv"), sep=";", index=False)
# df_weather_ap.to_csv(Path("../data/processed/data_weather_ap.csv"), sep=";", index=False)
```

```python
df_weather["AAAAMMJJ"] = pd.to_datetime(df_weather['AAAAMMJJ'].astype(str), format="%Y%m%d")
df_weather.rename(columns={'AAAAMMJJ':'Date'}, inplace=True)

df_weather_ap["AAAAMMJJ"] = pd.to_datetime(df_weather_ap["AAAAMMJJ"].astype(str), format="%Y%m%d")
df_weather_ap.rename(columns={"AAAAMMJJ": "Date"}, inplace=True)

df_weather_ap
```

```python
df_weather_ap.groupby("NUM_POSTE")[["PMERM", "GLOT", "UM"]].count()
```

```python
df_weather.groupby("NUM_POSTE")[["RR", "TNTXM", "FFM"]].count()
```

```python
df_weather = df_weather[["Date", "RR", "TNTXM", "FFM"]]
df_weather_ap = df_weather_ap[["Date", "PMERM", "GLOT", "UM"]]
df_weather_ap
```

```python
df_weather_ap.isnull().sum()
```

```python
df_weather_daily = df_weather.groupby("Date")[["RR", "TNTXM", "FFM"]].mean().reset_index()

df_weather_ap_daily = df_weather_ap.groupby("Date")[["PMERM", "GLOT", "UM"]].mean().reset_index()
df_weather_ap_daily
```

```python
# df_weather_daily.to_csv(Path("../data/processed/data_weather_filtered.csv"), sep=";", index=False)
# df_weather_ap_daily.to_csv(Path("../data/processed/data_weather_ap_filtered.csv"), sep=";", index=False)
```

## Merging the two weather datasets together

```python
df_weather_daily_full = pd.merge(df_weather_daily, df_weather_ap_daily, on="Date", how="outer")
df_weather_daily_full
```

# Merging dataset with weather variables

```python
df_o3 = pd.read_csv(Path("../data/processed/merged_o3_trafic.csv"), index_col=0)
df_pm10 = pd.read_csv(Path("../data/processed/merged_pm10_trafic.csv"), index_col=0)
```

```python
df_o3
```

```python
df_o3['date_fin'] = pd.to_datetime(df_o3['date_fin']) - pd.Timedelta(hours=1)
df_pm10['date_fin'] = pd.to_datetime(df_pm10['date_fin']) - pd.Timedelta(hours=1)
```

```python
# Creating new column date for merging
df_o3["Date"] = pd.to_datetime(df_o3["date_fin"]).dt.date
df_o3["Date"] = pd.to_datetime(df_o3["Date"])

df_pm10["Date"] = pd.to_datetime(df_pm10["date_fin"]).dt.date
df_pm10["Date"] = pd.to_datetime(df_pm10["Date"])
```

```python
df_o3
```

```python
# Merging df_o3 and df_pm10 with df_weather on "Date"

df_merged_o3 = pd.merge(df_o3, df_weather_daily_full, on="Date", how="left")
df_merged_o3.drop(columns=["Date"], inplace=True)

df_merged_pm10 = pd.merge(df_pm10, df_weather_daily_full, on="Date", how="left")
df_merged_pm10.drop(columns=["Date"], inplace=True)
```

```python
df_merged_o3.head(30)
```

```python
df_merged_o3.duplicated().sum()
```

```python
df_merged_o3["date"] = df_merged_o3["date_fin"].dt.floor("D")
df_merged_pm10["date"] = df_merged_pm10["date_fin"].dt.floor("D")
```

```python
# df_merged_o3.to_csv(Path("../data/processed/merged_o3_trafic_weather.csv"), index=False)
# df_merged_pm10.to_csv(Path("../data/processed/merged_pm10_trafic_weather.csv"), index=False)
```

# Creating dataset with daily observations

```python
# Group dataset by day and calculate mean for each day
df_merged_o3_daily = df_merged_o3.groupby(["date", "code_site"]).agg({
    "concentration": "max",
    "avg_debit": "mean",
    "avg_taux_occup": "mean",
    "RR": "mean",
    "TNTXM": "mean",
    "FFM": "mean",
    "PMERM": "mean",
    "GLOT": "mean",
    "UM": "mean"
}).reset_index()

df_merged_pm10_daily = (
    df_merged_pm10.groupby(["date", "code_site"])[["concentration", "avg_debit", "avg_taux_occup", "RR", "TNTXM", "FFM", "PMERM", "GLOT", "UM"]]
    .mean()
    .reset_index()
)
```
